### Orders "merge""
So, we decided that one order_id should stand for ONE order (and not for many rows like now). That's why we "collapsed" data together:
* we will keep sales (as a sum, one per order)
* we will also keepp amount of items per order (also through .transform() method)
* and the same for sales

We don't know yet if we will use this data or not but better to:
* have it in agregated version
* we will have unique order_id, which allows us to merge dataframes together with unique order_id

```python
# transform data: take together total sales, total profit, product quantity
df_orders['total_sales'] = df_orders.groupby("order_id")["sales"].transform("sum")
```

```python
# now profit:
df_orders['total_profit'] = df_orders.groupby("order_id")["profit"].transform("sum")
```

```python
# now quantity of the priducts:
df_orders['total_items'] = df_orders.groupby("order_id")["quantity"].transform("sum")
```

```python
df_orders.head(2)
```

```python
df_orders.head(2)
```

```python
# all country is United States --> we can drop the column
df_orders['country/region'].unique()
```

```python
df_orders = df_orders.drop(columns = 'country/region')
```

### Date Formats and new columns
Since we have the research focused on delivery dates / truck arrivals / warehouse open days, we need to have distribution by day (of a week) and also -- just in case -- by month and year (to see if there are some seasonal changes):

```python
# order date -> bring to several columns (year, month, day)
df_unique_orders["order_year"]=df_unique_orders["order_date"].dt.year
df_unique_orders["order_month"]=df_unique_orders["order_date"].dt.month
df_unique_orders["order_weekday"]=df_unique_orders["order_date"].dt.dayofweek
```
