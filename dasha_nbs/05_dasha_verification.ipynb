### On_truck_scan_date: validation
So far, I didn't really use the on_truck_scan_date to calculate any metric. Why:
* it doesn't show the details of the process, because it includes all the time from the order being registered to being picked up by a truck --> not really informative (e.g., we don't know where was the delay: by the warehouse or by the truck)

But! We can use it to validate our findings. 
Plus it's a larger dataset. 
So, here is the new metric: business days between the order registration and order being picked up.
* ```verify = on_truck_scan_date - order_date```

````python
import pandas as pd
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
````

````python
df_verify = pd.read_csv('../data_csv/orders_min.csv')
df_int = pd.read_csv('../data_csv/interns_min.csv')
df_proc = pd.read_csv('../data_csv/process_min.csv')
````

````python
df_verify.info()
````

````python
df_verify = df_verify.drop(columns=['Unnamed: 0','ship_mode','origin_channel', 'region', 'total_items'])
df_verify.info()
````

````python
df_proc.info()
````

````python
df_proc = df_proc.drop(columns=['Unnamed: 0', 'truck_ship_mode'])
df_proc.info()
````

````python
df_verify = df_verify.merge(df_proc, how = 'left', on='row_id')
df_verify.info()
````

````python
df_verify = df_verify.drop(columns=['row_id'])
df_verify.info()
````

````python
df_int.info()
````

````python
df_int = df_int.drop(columns=['Unnamed: 0'])
df_int.info()
````

````python
df_verify = df_verify.merge(df_int, how='left', on='order_id')
df_verify.info()
````

````python
df_ver_unique = df_verify.drop_duplicates(subset="order_id", keep="first")
````

````python
df_ver_unique.info()
````

````python
df_ver_unique["order_date"] = pd.to_datetime(df_ver_unique["order_date"])
df_ver_unique["ready_to_ship_date"] = pd.to_datetime(df_ver_unique["ready_to_ship_date"])
df_ver_unique["pickup_date"] = pd.to_datetime(df_ver_unique["pickup_date"])
df_ver_unique["on_truck_scan_date"] = pd.to_datetime(df_ver_unique["on_truck_scan_date"])
df_ver_unique.info()
````

Process accofing to warehouse: two metrics
* (warehouse) processing time: ```process_days_biz```
* waiting for the truck time: ```truck_wait_days_biz```

````python
# warehouse processing time, business days
mask = df_ver_unique["order_date"].notna() & df_ver_unique["ready_to_ship_date"].notna()

df_ver_unique.loc[mask, "process_days_biz"] = np.busday_count(
    df_ver_unique.loc[mask, "order_date"].values.astype("datetime64[D]"),
    df_ver_unique.loc[mask, "ready_to_ship_date"].values.astype("datetime64[D]")
)

df_ver_unique.info()
````

````python
# waiting for the truck days

mask = df_ver_unique["pickup_date"].notna() & df_ver_unique["ready_to_ship_date"].notna()

df_ver_unique.loc[mask, "truck_wait_days_biz"] = np.busday_count(
    df_ver_unique.loc[mask, "ready_to_ship_date"].values.astype("datetime64[D]"),
    df_ver_unique.loc[mask, "pickup_date"].values.astype("datetime64[D]")
)
df_ver_unique.info()
````

````python
# sum of the processing time (warehouse) + waiting for the truck days (warehouse)
# according to the interns / warehouse

# waiting for the truck days

# Using the + operator (NaN propagation)
df_ver_unique["warehouse_data"] = df_ver_unique["process_days_biz"] + df_ver_unique["truck_wait_days_biz"]

# Alternative: using sum with axis=1 (you can also ignore NaNs if you want)
# df_unique_all["warehouse_days_biz"] = df_unique_all[["process_days_biz", "truck_wait_days_biz"]].sum(axis=1, skipna=False)
df_ver_unique.info()

````

According to the information, coming from the warehouse sources, this is the descriptive statistics on the
* processing time
* waiting for the truck time:

````python
df_ver_unique["warehouse_data"].describe()
````

Now we calculate the same "estimation" using the data from the truck scans:
* ```truck_data = on_truck_scan_date - order_date```
* it should be the same as ```warehouse_data```

````python
# time between order creation and on_truck_scan_date:

mask = df_ver_unique["order_date"].notna() & df_ver_unique["on_truck_scan_date"].notna()

df_ver_unique.loc[mask, "truck_data"] = np.busday_count(
    df_ver_unique.loc[mask, "order_date"].values.astype("datetime64[D]"),
    df_ver_unique.loc[mask, "on_truck_scan_date"].values.astype("datetime64[D]")
)
df_ver_unique.info()
````
