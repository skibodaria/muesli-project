Column names

```python
df_intern.columns = df_intern.columns.str.lower()
df_intern.columns = df_intern.columns.str.replace(' ','_')
```

Deal with data format (surrently it is a string):

```python
df_intern["ready_to_ship_date"] = pd.to_datetime(df_intern["ready_to_ship_date"])
df_intern["pickup_date"] = pd.to_datetime(df_intern["pickup_date"])
```

Adding new columns with year / month / day / weekday (to track the shipment and delivery info):

```python
df_intern["ready_ship_year"]=df_intern["ready_to_ship_date"].dt.year
df_intern["ready_ship_month"]=df_intern["ready_to_ship_date"].dt.month
df_intern["ready_ship_weekday"]=df_intern["ready_to_ship_date"].dt.dayofweek

df_intern["pickup_year"]=df_intern["pickup_date"].dt.year
df_intern["pickup_month"]=df_intern["pickup_date"].dt.month
df_intern["pickup_weekday"]=df_intern["pickup_date"].dt.dayofweek
```

```python
# check the result:
df_intern.head(3)
```

Checking for duplicates

```python
# duplicates by order_id
df_intern["order_id"].duplicated().sum()
# >> 86 duplicates
```

```python
df_intern.duplicated().sum()
```

```python
# The duplicates we have are identical not only by the order_id but by the entire row content -> safe to drop them:
df_unique_intern = df_intern.drop_duplicates(subset="order_id", keep="first")
```

```python
df_unique_intern.shape
```

```python
df_unique_intern.duplicated().sum()
```

```python
df_unique_intern.to_csv('./data_csv/intern_unique.csv')
```
